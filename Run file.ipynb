```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("final_cleaned_output.csv")

# Optional: clean whitespace in column names
df.columns = df.columns.str.strip()

# Set Seaborn style
sns.set(style="whitegrid")

# Create pointplot: Bin vs. Average Return with 95% CI
plt.figure(figsize=(8, 5))
sns.pointplot(
    x="Bins",
    y="Return",
    data=df,
    estimator="mean",
    ci=95,
    join=True,
    capsize=0.2,
    color="darkgreen"
)

# Labels and title
plt.title("Average Return by 10-K Change Bin", fontsize=14)
plt.xlabel("10-K Change Bin (1 = Largest Change, 5 = Smallest)", fontsize=12)
plt.ylabel("Average Return (%)", fontsize=12)
plt.tight_layout()
plt.show()
```
